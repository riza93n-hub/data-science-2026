```python
import pandas as pd
import numpy as np

# 1. Muat Dataset Titanic dari URL GitHub
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)

# 2. Eksplorasi Awal Data
print("--- 5 Baris Pertama Dataset ---")
display(df.head())
print("\n--- Ukuran Dataset (Baris, Kolom) ---")
display(df.shape)
```

```python
# 1. Daftar nama kolom dalam tabel
print("Daftar nama kolom:")
for col in df.columns:
    print(f"- {col}")

print("\n" + "="*40 + "\n")

# 2. Mengecek kolom yang memiliki data kosong (Missing Values)
missing = df.isna().sum()
missing = missing[missing > 0]

if not missing.empty:
    max_len = max(len(col) for col in missing.index)
    print("Kolom dengan missing values:")
    for col, val in missing.items():
        print(f"- {col.ljust(max_len)} : {val}")
else:
    print("Tidak ada kolom dengan missing values.")
```

```python
# 1. Menyaring data penumpang wanita kelas 1 yang selamat
female_class1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
survived = female_class1[female_class1['Survived'] == 1]
survived_count = survived['Survived'].count()
avg_age = survived['Age'].mean()

print("Analisis Penumpang Wanita Kelas 1 yang Selamat:")
print(f"- Jumlah yang selamat : {survived_count} orang")
print(f"- Rata-rata usia      : {avg_age:.2f} tahun")

print("\n" + "="*40 + "\n")

# 2. Mencari kelas penumpang dengan tingkat keselamatan tertinggi
survival_rate = df.groupby('Pclass')['Survived'].mean()

print("Tingkat keselamatan per kelas:")
for pclass, rate in survival_rate.items():
    print(f"- Kelas {pclass} : {rate:.2%}")

best_class = survival_rate.idxmax()
best_rate = survival_rate.max()
print(f"\nKelas dengan tingkat keselamatan tertinggi adalah Kelas {best_class} ({best_rate:.2%})")
```

```python
print("=== HASIL JAWABAN PERTANYAAN EVALUASI ===")

# Q1: Berapa jumlah baris dan kolom dataset Titanic?
rows, cols = df.shape
print(f"Q1. Ukuran dataset Titanic -> Jumlah baris: {rows}, Jumlah kolom: {cols}")

# Q2: Kolom mana yang memiliki nilai kosong (missing)?
missing_cols = df.columns[df.isna().any()]
print(f"Q2. Kolom dengan missing values: {', '.join(missing_cols)}")

# Q3: Berapa persen penumpang yang selamat (Survived=1)?
overall_survival_rate = df['Survived'].mean() * 100
print(f"Q3. Persentase penumpang yang selamat: {overall_survival_rate:.2f}%")

# Q4: Rata-rata usia penumpang wanita kelas 1 yang selamat?
avg_age_female_c1 = df[
    (df['Sex'] == 'female') &
    (df['Pclass'] == 1) &
    (df['Survived'] == 1)
]['Age'].mean()
print(f"Q4. Rata-rata usia penumpang wanita kelas 1 yang selamat: {avg_age_female_c1:.2f} tahun")

# Q5 & Q6: Kelas penumpang mana yang tingkat keselamatan tertinggi beserta rincian kelompoknya?
survival_rate_group = df.groupby('Pclass')['Survived'].mean()
best_class_group = survival_rate_group.idxmax()

print(f"Q5. Kelas dengan tingkat keselamatan tertinggi: Kelas {best_class_group}")
print("Q6. Rincian hasil persentase keselamatan kelompok per kelas:")
for pclass, rate in survival_rate_group.items():
    print(f"    - Kelas {pclass}: {rate:.2%}")
```
